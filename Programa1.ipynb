```python
# Sistemas de préstamos
# Versión 1 - Progrmación básica y uso de listas
# ====================================

#Listas de alumnos activos en el sistema
alumnos=["26040200","26040201","26040202","2604203","2604204"]
# Cada valor en la lista de computadoras tiene su lista correspondiente con el
# aula donde se ubica, el software instalado y el estado de esa computadora.
computadoras=["PC-LC1-01","PC-LC1-02","PC-LC1-03",
              "PC-LC2-01","PC-LC2-02","PC-LC3-03",
              "PC-LC3-01","PC-LC3-02","PC-LC3-03","PC-LC3-04"]
aulas=["LC1","LC1","LC1",
       "LC2","LC2","LC2",
       "LC3","LC3","LC3","LC3"]
software=["Python, VS Code","Python, Java, VS Code","AutoCAD, MATLAB",
          "Python, VS Code","Python, Java, VS Code","AutoCAD, MATLAB",
          "Python, VS Code","Python, Java, VS Code","AutoCAD, MATLAB","Python, VS Code"]
estados=["Disponible", "Prestada","Reservada",
         "Disponible", "Prestada","Reservada",
         "Disponible", "Prestada","Reservada","Mantenimiento"]
# MENÚ PRINCIPAL
# Usamos un ciclo while que se repetira indefinidamente hasta que el usauario teclee
# la opcion 7 (salir) o cualquier numero superior a 7.
opcion=0
while opcion<=6:
  print("/n============")
  print("SISTEMA DE PRESTAMOS")
  print("1. Consultar Computadora")
  print("2. Buscar Softaware")
  print("3. Solicitar Préstamo")
  print("4. Devolver Computadora")
  print("5. Crear reservación")
  print("6. Agregar computadora")
  print("7. Salir")
  print("==============")
  opcion=int(input("Seleccione una opción: "))

# 1. Cosultar computadoras
# Si el usuario selecciono la opción 1 se hace un ciclo (for) que rocorre toda la lista
# de computadoras e imprime los valores de los demas listas.
  if opcion==1:
    print("\n--- COMPUTADORAS ---")
    for i in range(len(computadoras)):
      print("Aula:", aulas[i], "ID:", computadoras[i], "Softaware:", software[i], "Estado:", estados[i])

  # 2. Buscar Software
  elif opcion==2:
    programa=input("/ Ingrese el Software que desea buscar: ")
    print("/n--- Resultados ---")
    encontrado=False
    for i in range(len(computadoras)):

  # La funcion (lower) convierte a minuscula el texto, de esta forma nos aseguramos
  # que el texto a buscar y la información de las listas este en minuscula y no tengamos
  # problema de que no encuentra un programa por que este escritorio de forma diferente
  # usando y minusculas.
      if programa.lower()in software[i].lower():
        print("ID:", computadoras[i])
        print("Aula:", aulas[i])
        print("Software:", software[i])
        print("Estados:", estados[i])
        encontrado=True
    if encontrado==False:
      print("No se encontraron computadoras con ese software")

  #3. Solicitar préstamo
  elif opcion==3:
    print("\n --- Solicitar Préstamos ---")
    num_control=input("Ingrese el número de control")
    encontrado1=False
    # Verificar si existe el alumno en las lista o BD
    if num_control in alumnos:
      # Si existe, continuamos con el procedimiento.
      encontrado1=True
      id_computadora=input("Ingrese el ID de la computadora: ")
      encontrado2= False
      for i in range(len(computadoras)):
        if computadoras[i]==id_computadora:
          encontrado2=True
          if estados[i]=="Disponible":
            estados[i]="Prestada"
            print("Préstamo Registrado Correctamente.")
            print("Computadora:", computadoras[i])
            print("Aula", aulas[i])
          else:
              print("La computadora no esta disponible.")
              if encontrado2==False:
                print("La computadora no existe.")
              if encontrado1==False:
                print("El alumno no existe.")

#4. Devolver computadora
  elif opcion==4:
    print("\n --- Devolver Computadora ---")
    id_pc=input("Dame el ID de la Computadora ")
    encontrado=False
    for i in range(len(computadoras)):
      if computadoras[i]==id_pc:
        encontrado=True
        if estados[i]=="Prestada" or estados[i]=="Mantenimiento":
          estados[i]="Disponible"
          print("La computadora", id_pc, "Ahora esta disponible")
        else:
          print("Esa Computadora No Esta Prestada o Reservada")
    if encontrado==False:
      print("No se encontro ese ID")

#6. Agregar computadora
  elif opcion==6:
    print("\n --- AGREGAR COMPUTADORA ---")
    id_computadora=input("ID de la computadora: ")
    aula=input("Aula: ")
    programas=input("Software Instalado: ")
    computadoras.append(id_computadora)
    aulas.append(aula)
    software.append("Disponible")
    print("\n Computadora Agregada Correctamente. ")

```
